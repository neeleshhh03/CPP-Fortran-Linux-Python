```python
#CONDITIONAL STATEMENTS
name= str(input('Enter you full name: '))
marks= float(input('Enter your marks(out of 100): '))

if(marks>= 85.0):
    print(name, ' has achieved A grade.')
if(marks< 85.0 and marks>=60.0):
    print(name, ' has achieved B grade.')
if(marks< 60.0 and marks>=40.0):
    print(name, ' has achieved C grade.')
else:
    print(name, ' NIGGA is failed.')
```
