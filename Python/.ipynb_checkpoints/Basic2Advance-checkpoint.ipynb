```python
#List Methods
list=[2,1,3]
list.append(4)   #mutated
print(list)

list.sort()   #returns None
print(list.sort())
print(list)

list.sort(reverse=True)
print(list)

print("=========")
list.insert(0, 433)   #0. index    433. element
print(list)

list.remove(3)   #removed 1st occcurance of 3
list.pop(3)       #removed index 3's value= 1
print(list)
```
