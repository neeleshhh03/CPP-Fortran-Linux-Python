```python
#Prime Number

num= int(input("enter a number: "))

if (num<= 1):
    print(num, "isn't a prime number")
else:
    for i in range(1, num-1, 1):
        if (num%i==0):
            print(num, "is a composite number")
            break
        else:
            print(num, 'is a prime number')
            break
```

```python
# WAP to calculate tax
salary = int(input('Enter your gross income Rs. '))
city = int(input('Enter your job city tier(1 or 2 or 3): '))

print()
print('------------------------------')
# HRA calculation
if (city == 1):
    hra = 50
    HRA = salary * (hra / 100)
    print("your HRA is Rs. ", HRA)
elif (city == 2):
    hra = 40
    HRA = salary * (hra / 100)
    print("Your HRA is Rs. ", HRA)
else:
    HRA = 30
    print("Your HRA is Rs. ", HRA)

# Direct Tax Calculation 
# [Total taxable income % bracket]= Taxable amount - Initial amount of % bracket
taxable = salary - 75000
print('Your Taxable income amt. is Rs.', taxable)

# Fixed: India uses progressive tax slabs. 
# Tax is calculated on parts of your income falling in specific slabs, not the whole amount.
tax = 0

if taxable <= 400000:
    tax = 0
elif taxable <= 800000:
    tax = (taxable - 400000) * 0.05
elif taxable <= 1200000:
    tax = (400000 * 0.05) + (taxable - 800000) * 0.10
elif taxable <= 1600000:
    tax = (400000 * 0.05) + (400000 * 0.10) + (taxable - 1200000) * 0.15
elif taxable <= 2000000:
    tax = (400000 * 0.05) + (400000 * 0.10) + (400000 * 0.15) + (taxable - 1600000) * 0.20
elif taxable <= 2400000:
    tax = (400000 * 0.05) + (400000 * 0.10) + (400000 * 0.15) + (400000 * 0.20) + (taxable - 2000000) * 0.25
else:
    tax = (400000 * 0.05) + (400000 * 0.10) + (400000 * 0.15) + (400000 * 0.20) + (400000 * 0.25) + (taxable - 2400000) * 0.30

# Calculate 4% Health and Education Cess on the base tax
cess_tax = tax * 0.04
total_tax = tax + cess_tax
print('Tax amount is Rs.', total_tax)
print('------------------------------')
```
